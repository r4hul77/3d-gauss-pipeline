```python
import cv2
def revec_tvec2T(rvec, tvec):
    R, _ = cv2.Rodrigues(rvec)
    T = np.eye(4)
    T[:3, :3] = R
    T[:3, 3:4] = -R@tvec
    return T

def invert_T(T):
    R = T[:3, :3]
    t = T[:3, 3]
    Ti = np.eye(4, dtype=np.float64)
    Ti[:3, :3] = R.T
    Ti[:3, 3] = -R.T @ t
    return Ti

imgs = [cv2.imread(img_path) for img_path in obj_file['good_imgs']]
Hs = obj_file['Hs']
Ts = [revec_tvec2T(rvec, tvec) for rvec, tvec in zip(obj_file['rvec'], obj_file['tvec'])]
K = obj_file['K']
```

```python
import numpy as np
def mask_under_template_dense(
    img,
    H_template_to_img,
    template_wh,
    H_is_template_to_img=True,
):
    Wt, Ht = template_wh

    if H_is_template_to_img:
        H_i2t = np.linalg.inv(H_template_to_img)
    else:
        H_i2t = H_template_to_img

    # Warp image into template frame
    warped = cv2.warpPerspective(
        img,
        H_i2t,
        (Wt, Ht),
        flags=cv2.INTER_NEAREST,
        borderValue=0
    )

    # Mask: valid pixels are non-zero after warp
    if warped.ndim == 3:
        mask = np.any(warped != 0, axis=2)
    else:
        mask = warped != 0

    return warped, mask
```

```python
def normalizeH(H, K):
    return np.linalg.inv(K)@H@K
normalizeH(Hs[0], K)
```

```python
def plane_eqn_From_homography_pose(H, K, T, eps=1e-9):
    Hn = normalizeH(H, K)
    R = T[:3, :3]
    M = Hn - R
    t = T[:3, 3:4]
    denom = float((t.T@t)[0, 0])
    if denom < eps:
        print("Zero Translation !!")
        raise ValueError("Zeros Translation")
    wT = (t.T @ M)/denom
    w = wT.reshape(3)
    nw = np.linalg.norm(w)
    if nw < eps:
        raise ValueError("Degen Plane !")
    n = w / nw
    d = 1.0 / nw

    if d < 0:
        n = -n
        d = -d
    return np.hstack([n ,d ])
    
def rel_pose_c0_to_c1(T_w_c0, T_w_c1):
    # T_10 = T_c1_w * T_w_c0
    T_c1_w = invert_T(T_w_c1)
    return T_c1_w @ T_w_c0
    
def project(K, X):
    # X: (N,3) in camera coords
    x = (K @ X.T).T
    return x[:, :2] / x[:, 2:3]

def apply_H(H, x0):
    # x0: (N,2) pixels in img0
    x0_h = cv2.convertPointsToHomogeneous(x0).reshape(-1,3)
    y = (H @ x0_h.T).T
    return y[:, :2] / y[:, 2:3]
```
